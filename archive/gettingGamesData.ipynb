```python
from bs4 import BeautifulSoup
import pandas as pd
pd.set_option('display.max_rows', None)
import requests
```

```python
url = "https://en.wikipedia.org/wiki/List_of_best-selling_video_games"
header = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
response = requests.get(url, headers=header)
soup = BeautifulSoup(response.content, "html.parser")
```

```python
allTables = soup.find_all("table", {"class": "wikitable"})
gameTable = allTables[0]

#print(gameTable.prettify)
gameData = {}
for row in gameTable.find_all("tr")[1:]:
    idx = 1
    gameName = row.find("th").text.strip("\n")
    sales = row.find_all("td")[idx].text.strip("\n")
    idx += 3
    if sales.isalpha():
        sales = row.find_all("td")[0].text.strip("\n")
        idx-=1
    releaseYear = row.find_all("td")[idx].text.strip("\n")
    print(f"{gameName} sold {sales} in {releaseYear}")
```

```python
url = "https://en.wikipedia.org/wiki/List_of_best-selling_books"
header = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
response = requests.get(url, headers=header)
soup = BeautifulSoup(response.content, "html.parser")
```

```python
allTables = soup.find_all("table", {"class": "wikitable"})
copies1M =  allTables[0]
bestBooks = []

for row in copies1M.find_all("tr")[1:]:
    title = row.find_all("td")[0].text.strip()
    author = row.find_all("td")[1].text.strip()
    year = row.find_all("td")[3].text.strip()
    sales = row.find_all("td")[4].text.strip().split("million")[0]
    genre = row.find_all("td")[5].text.strip()
    if not sales[0].isnumeric():
        sales = sales[1:]
    sales = int(sales)
    bestBooks.append({"Title": title, "Author": author, "Year": year, "Sales": sales, "Genre": genre})
    
print(bestBooks)
```

```python
bestSellingBooks = pd.DataFrame(bestBooks)
bestSellingBooks
```

```python
def addToDataFrame(tableNum):
    global bestSellingBooks
    copies1M =  allTables[tableNum]
    bestBooks = []

    for row in copies1M.find_all("tr")[1:]:
        title = row.find_all("td")[0].text.strip()
        author = row.find_all("td")[1].text.strip()
        year = row.find_all("td")[3].text.strip()
        sales = row.find_all("td")[4].text.strip().split("million")[0]
        try:
            genre = row.find_all("td")[5].text.strip()
        except:
            genre = "N/A"
        if not sales[0].isnumeric():
            sales = sales[1:]
        try:
            sales = float(sales)
        except:
            sales = sales
        
        bestBooks.append({"Title": title, "Author": author, "Year": year, "Sales": sales, "Genre": genre})

    temp = pd.DataFrame(bestBooks)
    bestSellingBooks = pd.concat([bestSellingBooks, temp], axis = 0, join = "outer", ignore_index=True)
```

```python
for n in range(1, 3):
    addToDataFrame(n)
```
